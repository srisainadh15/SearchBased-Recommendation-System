```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
df = pd.read_csv('final_df.csv')
```

```python
df.head()
```

```python
df = df.drop_duplicates(subset=['Book-Title']).reset_index(drop= False)
df
```

```python
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(df['Book-Title'])
```

```python
tfidf_matrix.toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

 
```

```python
def search_based_recommendation(book_title,df,tfidf,tfidf_matrix,n=10):
    query_vector = tfidf.transform([book_title])
    cosin_sim  = cosine_similarity(query_vector,tfidf_matrix).flatten()
    sim_scores = list(enumerate(cosin_sim))
    sim_scores = sorted(sim_scores,key=lambda x:x[1],reverse=True)
    sim_scores = sim_scores[1:n+1]
    book_indices = [i[0] for i in sim_scores]
    return df[['Book-Title','Book-Rating','Image-URL-M']].iloc[book_indices].reset_index(drop=True)
    
   
```

```python
df[['Book-Title','Book-Rating']].iloc[4]
```

```python
book_to_search = "The Fountainhead"
recommend_books = search_based_recommendation(book_to_search,df,tfidf,tfidf_matrix)
recommend_books
```

```python
book_to_search = "Pinocchio"
recommend_books = search_based_recommendation(book_to_search,df,tfidf,tfidf_matrix)
recommend_books
```
